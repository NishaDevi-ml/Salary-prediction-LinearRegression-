```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Salary=pd.read_csv("Salary_Data.csv")
Salary
```

```python
X=Salary[["YearsExperience"]]
Y=Salary["Salary"]
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(X,Y)
```

```python
m=lr.coef_
c=lr.intercept_
m
```

```python
y=m*11+c
print(y)
```

```python
plt.scatter(X,Y)
plt.xlabel("Experience")
plt.ylabel("salary")
plt.plot(X.values,lr.predict(X))
plt.show()
```
